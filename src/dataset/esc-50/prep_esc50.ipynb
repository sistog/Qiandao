```python
import os 
import numpy as np
import librosa
import soundfile as sf
dataset_path = "/data/zcx/wav_prj/HTS-Audio-Transformer-main/esc-50/ESC-50-master"
meta_path = os.path.join(dataset_path, "meta", "esc50.csv")
audio_path = os.path.join(dataset_path, "audio")
resample_path = "/data/zcx/wav_prj/HTS-Audio-Transformer-main/esc-50/ESC-50-master/audio_16k/"
```

```python
# prepare for the folder
meta = np.loadtxt(meta_path , delimiter=',', dtype='str', skiprows=1)
audio_list = os.listdir(audio_path)
```

```python
# 确保输出目录存在
os.makedirs(resample_path, exist_ok=True)

print(f"开始重采样 {len(audio_list)} 个文件...")

for i, f in enumerate(audio_list):
    full_f = os.path.join(audio_path, f)
    resample_f = os.path.join(resample_path, f)
    
    try:
        # 使用librosa重采样
        y, sr = librosa.load(full_f, sr=None)
        y_resampled = librosa.resample(y, orig_sr=sr, target_sr=16000)
        sf.write(resample_f, y_resampled, 16000)
        
        if i % 50 == 0:  # 每50个文件打印一次进度
            print(f"已处理 {i+1}/{len(audio_list)} 个文件")
            
    except Exception as e:
        print(f"处理文件 {f} 时出错: {e}")

print("重采样完成！")
```

```python
# name fold target category esc10 srcfile take
output_dict = [[] for _ in range(5)]
for label in meta:
    name = label[0]
    fold = label[1]
    target = label[2]
    y, sr = librosa.load(os.path.join(resample_path, name), sr = None)
    output_dict[int(fold) - 1].append(
        {
            "name": name,
            "target": int(target),
            "waveform": y
        }
    )
    
```

```python
np.save("esc-50-data.npy", output_dict)
```
